# Hands On Data Lab Implementation
This notebook demonstrates data loading, inspection, cleaning, summary statistics, and visualization using Python, pandas, and matplotlib.

The included student study dataset is fictional and is only for practice.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('student_study_data.csv')
df.head()
```

## 1. Inspect the data
Check the number of rows and columns, data types, and missing values.

```python
print('Shape:', df.shape)
print('\nData types:')
print(df.dtypes)
print('\nMissing values:')
print(df.isnull().sum())
print('\nSummary:')
display(df.describe())
```

## 2. Data manipulation
Create a category based on study hours, sort by score, and filter students scoring at least 80%.

```python
df['study_level'] = df['study_hours'].apply(lambda hours: 'High' if hours >= 3 else 'Regular')
top_students = df[df['score_percent'] >= 80].sort_values('score_percent', ascending=False)
display(top_students)
display(df.groupby('study_level')[['study_hours', 'score_percent']].mean())
```

## 3. Data visualization
Make a bar chart of student scores and a scatter plot comparing study hours with scores.

```python
plt.figure(figsize=(9, 5))
plt.bar(df['student'], df['score_percent'])
plt.title('Practice Scores by Student')
plt.xlabel('Student')
plt.ylabel('Score (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(7, 5))
plt.scatter(df['study_hours'], df['score_percent'])
plt.title('Study Hours and Practice Scores')
plt.xlabel('Study hours')
plt.ylabel('Score (%)')
plt.tight_layout()
plt.show()
```

## 4. Basic analysis
Calculate the average score and study time, then describe the visible relationship.

```python
print('Average score:', round(df['score_percent'].mean(), 2), '%')
print('Average study time:', round(df['study_hours'].mean(), 2), 'hours')
print('Highest score:', df['score_percent'].max(), '%')
print('Lowest score:', df['score_percent'].min(), '%')
print('Correlation:', round(df['study_hours'].corr(df['score_percent']), 2))
```

## Conclusion
In this small fictional dataset, students with more study hours generally tend to have higher practice scores. This is an observed association in a small sample, not proof that study time alone causes a higher score. More data would be needed for a stronger conclusion.
